```python
import json
import numpy as np
```

```python
def load_data(filename) :
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
data
```

```python
# cleaning data --

def clean_data(data) :
    text_to_num = {"one" : 1, "two" : 2, "three" : 3, "four"  : 4, "five" : 5}
    cleaned_data = []
    uniq_user = set()

    for user in data :
        # rating -
        str_rating = str(user["rating"]).strip().lower()
        if str_rating in text_to_num :
            str_rating = text_to_num[str_rating]
        user["rating"] = str_rating

        # missing val -
        raw_age = user.get("age")
        if raw_age == None :
            user["age"] = None
        
        # duplicate data -
        if user["name"].strip() in uniq_user  :
            continue

        uniq_user.add(user["name"])
        cleaned_data.append(user)
        
    return cleaned_data
```

```python
cleaned_data = clean_data(data)
```
