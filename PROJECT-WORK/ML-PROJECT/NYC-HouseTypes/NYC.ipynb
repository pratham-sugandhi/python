```python
# Data Cleaning =>

# drop cols
df_clean = df.drop(columns=["id", "name", "host_name", "last_review"])

# No reviews yet 
df_clean["reviews_per_month"] = df_clean["reviews_per_month"].fillna(0)
```

```python
# Skew =>
df_clean[numerical_cols].skew()
```

```python

# Cap outlier -upper
price_cap = df_clean["price"].quantile(0.99)
nights_cap = df_clean["minimum_nights"].quantile(0.99)

df_clean["price"] = df_clean["price"].clip(upper=price_cap)
df_clean["minimum_nights"] = df_clean["minimum_nights"].clip(upper=nights_cap)


"""
df_clean = df_clean[df_clean["price"] < price_cap]
df_clean = df_clean[df_clean["minimum_nights"] < nights_cap]
"""

df_clean.head()
```

# Train Test =>


```python

X = df_clean.drop(columns=["room_type"])
y = df_clean["room_type"]
```

```python
# stratify => classifi : split main - group / class ka proportion maintain rahe

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.33, stratify=y
)

```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer
from sklearn.compose import ColumnTransformer
```

```python
# Pipeline =>

numerical_cols = ["price", "minimum_nights", "number_of_reviews", "reviews_per_month",
                  "calculated_host_listings_count", "availability_365", "latitude", "longitude"]

categorical_cols =["neighbourhood_group", "neighbourhood"]

# 1 num =>
numeric_pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="median")),
    # ("skewness", PowerTransformer(method="yeo-johnson")),
    ("scale", StandardScaler())
])

# 2 cat =>
categorical_pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("encode", OneHotEncoder(handle_unknown="ignore"))
])
```

```python

# ColumnTransformer =>
preprocessor = ColumnTransformer(transformers=[
    ("numerical", numeric_pipeline, numerical_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

preprocessor
```

```python
# Algortihm =>
# SMOTE for class IMBalance -> fake val
"""
Overfitting 
Data Leaks
"""


# Class Weights ==> more focus on less data

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
```

```python
# (class_weight="balanced") => pay more attention to minority class in case of skewness
# GB doesnt support class_weight

models = {
    # name : model
    "Logistic Regression" : LogisticRegression(class_weight="balanced", random_state=42),
    "Decision Tree" : DecisionTreeClassifier(class_weight="balanced", random_state=42),
    "Random Forest" : RandomForestClassifier(class_weight="balanced", random_state=42),
    "Gradient Boosting" : GradientBoostingClassifier(random_state=42)
}

for name, model in models.items():
    pipe = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    # acc = misleading due to imbalanced/skewed class, so use macro f1
    
    accuracy = cross_val_score(pipe, X_train, y_train, cv=3, scoring="accuracy")
    macro_f1 =  cross_val_score(pipe, X_train, y_train, cv=3, scoring="f1_macro")

    print(f"{name} -> acc: {accuracy.mean():.3f}, F1: {macro_f1.mean():.3F}")
    
    
```

# Hyper parameter Tuning

```python
from sklearn.model_selection import RandomizedSearchCV

best_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier",  RandomForestClassifier(class_weight="balanced"))
])

param_distribution = {
    "classifier__n_estimators" : [100,200,150,300],
    "classifier__max_depth" : [8, 12, 15, 20, None],
    "classifier__min_samples_split" : [2, 5 ,10]
}

search = RandomizedSearchCV(
    estimator = best_pipeline,
    param_distributions = param_distribution,
    n_iter = 10,
    cv = 3,
    scoring = "f1_macro",
    random_state = 42
)

search.fit(X_train, y_train)

```

```python

print("Best Parameters :", search.best_params_)
print("Best CV Macro-f1 :", search.best_score_)
```

# EValuation 

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score

best_pipeline = search.best_estimator_
y_pred = best_pipeline.predict(X_test)

print(f"Acc score: {accuracy_score(y_test, y_pred)}")
print(f"F1 score: {f1_score(y_test, y_pred, average="macro")}")
```

```python

sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt=".2f",
           xticklabels=best_pipeline.classes_ , yticklabels=best_pipeline.classes_)
```
