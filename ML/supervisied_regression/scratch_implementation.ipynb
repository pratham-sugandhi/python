```python
import numpy as np
```

```python
# Linear Regression --

class LinearRegression :
    def __init__(self, learning_rate= 0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def fit(self, X, y):
        m, n = X.shape #num of samples, features

        # step - 1 : initialize random val
        self.bias = 0
        self.weights = np.zeros(n)


        for i in range(self.n_iter) :
            # step - 2 : y_pred cost fun
            y_pred = self.bias + np.dot(X, self.weights) #0o + (01.x1 , 02.x2, ...)
    
            # step - 3 : derivative 
            db = (1/m) * np.sum(y_pred - y)
            dw = (1/m) * np.dot(X.T , (y_pred - y))  
            #dot becz data is stored in matrix , x ko trandpose kiya 
    
            # step - 4 : convergence
            self.bias -= self.lr* db
            self.weights -= self.lr* dw

    def predict(self, X) :
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([ [1], [2], [3], [4], [5] ])
y = np.array([2,4,6,8,10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
```

```python
# Linear Regression with OLS --

class LinearRegressionOLS:
    def __init__(self):
        self.bias = None
        self.weights = None
        
    def fit(self, X, y):
        m , n = X.shape

        X_b = np.c_[np.ones((m, 1)), X]

        #normal eqn
        # theta = np.dot( np.linalg.inv(np.dot(X_b.T, X_b)), np.dot(X_b.T, y))
        theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
        
        self.bias = theta[0]
        self.weights = theta[1:]
        
    def predict(self, X) :
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([ [1], [2], [3], [4], [5] ])
y = np.array([2,4,6,8,10])

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
print(model.bias)
print(model.weights)
```

```python
# Logistic Regression --

class LogisticRegression :
    def __init__(self, learning_rate= 0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def _sigmoid(self, z) :
        return (1/ (1+ np.exp(-z)) )

    def fit(self, X, y):
        m, n = X.shape #num of samples, features

        # step - 1 : initialize random val
        self.bias = 0
        self.weights = np.zeros(n)

        for i in range(self.n_iter) :
            z = self.bias + np.dot(X, self.weights)
            y_pred = self._sigmoid(z)

            # step - 3 : derivative 
            db = (1/m) * (y_pred - y)
            dw = (1/m) * np.dot(X.T , (y_pred - y))  
            #dot becz data is stored in matrix , x ko trandpose kiya 
    
            # step - 4 : convergence
            self.bias -= self.lr* db
            self.weights -= self.lr* dw

    def get_prob(self,  x):
            z = self.bias + np.dot(X, self.weights)
            return self._sigmoid(z)
        # in btw 0 - 1
        
    def predict(self, X, threshold = 0.5) :
        prob = self.get_prob(X)
        y_pred_bool = prob >= threshold
        return y_pred_bool.astype(int)
        
```

```python
# KNN CLASSIFIER --

class KNNClassifier :
    def __init__(self, k=3) :
        self.k = k

    def _euclidean_dist(self, x1, x2):
        return np.sqrt(np.sum((x1-x2)**2))

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def _predict_one (self, x):
        dist = [self._euclidean_dist(x, X_train) for x_train in self.X_train]

        knn_indices = np.argsort(dist)[:self.k]  #closest num ke index pos

        knn_classes = [self.y_train[i] for i in knn_indices]

        majority_class = np.argmax(np.bincount(knn_classes))  
        #bincount stores - freq of classes
        #argmax return index with max frequency 

        return majority_class

    def predict(self, X):
        y_pred = [self._predict_one(x) for x in X]
        return np.array(y_pred)

        
```

```python
# KNN --

X_train = np.array([
    [1,2],
    [2,3],
    [3,4],
    [4,5],
    [6,5],
    [7,7]
])

y_train = np.array([0,0,0,1,1])

X_test = np.array([
    [2,2],
    [6,6]
])

model = KNNClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(y_pred)
```
