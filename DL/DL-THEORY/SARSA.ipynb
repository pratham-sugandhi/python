```python
import gymnasium as gym
import numpy as np
```

```python
import random

env = gym.make("CliffWalking-v1")
```

```python
print(env.observation_space.n) # states
print(env.action_space.n) #action

starting_state , _ = env.reset()  #36 = 3 * 12 + 0
print(starting_state)
```

```python
## SARSA -->
# formula -
gamma = 0.99
alpha = 0.5
epsilon =  0.1
episodes = 500
```

```python
# Q- table : q values ( action -> feature reward)

Q = np.zeros((48,4))
```

```python
# Policy - epsilon-greedy : state -> action

# def epsilon_greedy(state):
#     if random.random() < epsilon :
#         # random action => explore
#         env.action_space.sample()
#     else :
#         # exploit
#         return np.argmax(Q[state])

def epsilon_greedy(state):
    if random.random() < epsilon:
        # Explore
        return env.action_space.sample()
    else:
        # Exploit
        return int(np.argmax(Q[state]))
        
```

```python

for episode in range(episodes) :
    render = (episode % 50 == 0)

    if render : 
        env = gym.make("CliffWalking-v1", render_mode = "human")
        print(f"rendering.. ep = {episode+1/500}")
    else :
        env = gym.make("CliffWalking-v1")

    done = False # when episode has ended
    state, _ = env.reset()
    action = epsilon_greedy(state)

    total_reward = 0
    episode_len = 0

    while not done :
        next_state, reward , terminated, truncated, _= env.step(action)
        done = terminated or truncated
        next_action = epsilon_greedy(next_state)

        # SARSA -
        Q[state, action] += alpha * (reward + gamma*(Q[next_state, next_action] - Q[state, action]))

        state = next_state
        action = next_action

        total_reward += reward
        episode_len += 1

    print(f" tot reward = {total_reward} &  len = {episode_len}")
    env.close()
    
```

```python
# What did agent learn"?

env = gym.make("CliffWalking-v1", render_mode="human")
state, _ = env.reset()
done = False
total_reward = 0
episode_len = 0

while not done :
    action = np.argmax(Q[state])
    state, reward , terminated, truncated, _= env.step(action)
    done = terminated or truncated

    total_reward += reward
    episode_len += 1

print(f"tot reward = {total_reward} & len {episode_len}")
env.close()
```
