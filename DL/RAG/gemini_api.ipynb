```python
# It lets you integrate these generative models into your applications
# to generate text and images, analyze multimodal inputs, and build conversational agents.

# Prompt to gemini model and get response

from google import genai

client = genai.Client(api_key = GEMINI_API_KEY)

interaction = client.interactions.create(
    model="gemini-3.8-flash",
    input="Explain how AI works in a few words"
)

print(interaction.output_text)
```

```python
# Image =>

uploaded_file = client.files.upload(file="Flag_of_India.svg.webp")

interaction = client.interactions.create(
    model="gemini-3.8-flash",
    input=[
        {"type": "text", "text": "which country's flag is this"},
        {
            "type": "image",
            "uri": uploaded_file.uri,
            "mime_type": uploaded_file.mime_type
        }
    ]
)
print(interaction.output_text)
```

```python
import os
from google import genai


class PersonalAssistant:

    def __init__(self):
        self.client = genai.Client(api_key=GEMINI_API_KEY)
        self.model = "gemini-3.8-flash"

        print("Hi, I'm your personal assistant!")

    def ans_query(self):
        query = input("Enter your query: ")

        interaction = self.client.interactions.create(
            model=self.model,
            input=query,
            temperature = 0.3,
            max_output_tokens = 512
        )

        print("\nAssistant:")
        print(interaction.output_text)

    def summarize_email(self):
        email = input("\nPaste your email:\n")

        interaction = self.client.interactions.create(
            model=self.model,
            input=f"""
            Summarize the following email.

            Include:
            - Main point
            - Important details
            - Action items
            - Deadlines

            Keep the summary concise.

            Email:
            {email}
            """
        )

        print("\nEmail Summary:")
        print(interaction.output_text)


assistant = PersonalAssistant()
```

```python

# Take query from user
assistant.summarize_email()
```
